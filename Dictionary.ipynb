```python
#1. Write a program to add a key and value to a dictionary.
my_dict={0:10,1:20}
my_dict[2]=30
print("Dictionary:",my_dict)
```

```python
#2. Write a program to concatanate the following dictionaries to create a new one.
dict1={1:10,2:20}
dict2={3:30,4:40}
dict3={5:50,6:60}
new_dict={}
new_dict.update(dict1)
new_dict.update(dict2)
new_dict.update(dict3)
print("New Dictionary",new_dict)
```

```python
#3 Write a program to Check if a given key already exists in a dictionary
my_dict={101:"Amit",102:"Rahul",103:"Manan"}
key=int(input("Enter a key to search:"))
if key in my_dict:
    print("Key exists in the Dictionary")
else:
    print("Key does not exist in the Dictionary")
```

```python
#4 Write a progam to iterate over a dictionary using for loop and print the keys alone,values alone and both keys and values
my_dict={
    101:"Amit",
    102:"Rahul",
    103:"Manan"
}
print("Keys:")
for key in my_dict:
    print(key)
print("\nValues:")
for value in my_dict.values():
    print(value)
print("\nKeys and Values:")
for key,value in my_dict.items():
    print(key,value)
```

```python
#5 Write a program to prepare a dictionary where the keys are numbers between 1 and 15(Incl) and the values are the square of the keys.
my_dict={}
for i in range(1,16):
  my_dict[i]=i*2
print(my_dict)
```

```python
#6 Write a program to sum all the values in a dictionary, considering the value will be of int type
my_dict={"a":10,"b":20,"c":30,"d":40}
total=0
for value in my_dict.values():
    total=total+value
print("Sum of Values:",total)
```

```python
#1. Write a program to print the 4th element from the first and 4th element from last in a tuple
numbers=(10,20,30,40,50,60,70,80,90)
print("Tuple:", numbers)
print("4th element from first:",numbers[3])
print("4th element from last:",numbers[-4])
```

```python
#2. Write a program to check wheather an element exists in a tuple or not.
num=(10,20,30,40,50)
element=int(input("Enter element to search:"))
if element in num:
    print("Element exists in the tuple")
else:
    print("Element does not exist in tuple")
```

```python
#3. Write a program to convert a list into tuple 
numbers_list=[10,20,30,40,50]
numbers_tuple=tuple(numbers_list)
print("List:",numbers_list)
print("Tuple:",numbers_tuple)
```

```python
#4. Write a Program to find the index of an item in a tuple
num=(10,20,30,40)
element=int(input("Enter Element:"))
if element in num:
    print("Index of element:",numbers.index(element))
else:
    print("Element not found")
```

```python
#5. Write a program to replace last value of tuples in a list to 100.
num=[(10,20,30),(40,50,60),(70,80,90)]
new_list=[]
for tup in num:
    new_tup=tup[:-1]+(100,)
    new_list.append(new_tup)
print("Orignal List:",num)
print("Updated List:",new_list)
```

```python
#6. Python program to find the maximum and minimum element in a tuple
numbers=(25,10,50,3,20,60)
maximum=max(numbers)
minimum=min(numbers)
print("Tuple",numbers)
print("Maximum elements:",maximum)
print("Minimum elements:",minimum)
```

```python
#7 Python program to swap the first and last element of a tuple.
num=(10,20,30,40,50)
new_tuple=(num[-1],)+num[1:-1]+(num[0],)
print("Orignal Tuple",num)
print("Tuple after swapping",new_tuple)
```

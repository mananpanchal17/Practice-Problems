```python
#Function Arguments
def evenodd(x:int)-> str:
    if(x%2==0):
       return "Even"
    else:
        return"Odd"
print(evenodd(17))
print(evenodd(6))
```

```python
#Default Arguments
def myFun(x,y=50):
    print("x",x)
    print("y",y)
myFun(10)
   
```

```python
def my_function(country="Norway"):
    print("I am from"+country)
my_function("Sweden")
my_function("India")
my_function()
my_function("Brazil")
```

```python
#Key Arguments
def student(fname,lname):
    print(fname,lname)
student(fname="Good",lname="Practice")
student(fname="Practice",lname="Good")
```

```python
#Postional Arguments
def nameAge(name,age):
    print("Hi,I am",name)
    print("My Age is",age)
nameAge("Suraj",27)
nameAge(27,"Suraj")
```

```python
#Arbitary Keyword
def myFun(*argv):
    for arg in argv:
        print(arg)
myFun('Hello','Welcome','to','MIT')
```

```python
#Recursion
def fact(n):
    if n==0:
        return 1
    else:
        return n*fact(n-1)
print(fact(4))
```

```python
#9 Accept 20 Values from user and save it in List. Perform Following
# Simple program to accept 20 values and perform operations on the list

values = []

# Accept 20 values from user
print("Enter 20 numbers:")
for i in range(20):
    num = int(input("Enter number: "))
    values.append(num)

print("\nList is:", values)

# a) Count similar elements and print their index values
print("\na) Similar elements and their index:")
checked = []
for i in range(len(values)):
    if values[i] not in checked:
        count = values.count(values[i])
        if count > 1:
            print("Value", values[i], "is repeated", count, "times")
            for j in range(len(values)):
                if values[j] == values[i]:
                    print("Found at index", j)
        checked.append(values[i])

# b) Count even and odd values
even = 0
odd = 0
for i in range(len(values)):
    if values[i] % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1

print("\nb) Even numbers:", even)
print("Odd numbers:", odd)

# c) Count positive and negative values
positive = 0
negative = 0
for i in range(len(values)):
    if values[i] > 0:
        positive = positive + 1
    elif values[i] < 0:
        negative = negative + 1

print("\nc) Positive numbers:", positive)
print("Negative numbers:", negative)
```

```python
#14 To Insert,delete and display
# Simple program to insert, delete, and display elements of a list
values = []
 
# Accept 5 elements from user using for loop
print("Enter 5 numbers:")
for i in range(5):
    num = int(input("Enter number: "))
    values.append(num)
 
print("\nOriginal list:", values)
 
# Insert an element at a given position
item = int(input("\nEnter element to insert: "))
pos = int(input("Enter position to insert at: "))
values.insert(pos, item)
print("List after insertion:", values)
 
# Delete an element
item = int(input("\nEnter element to delete: "))
if item in values:
    values.remove(item)
    print("List after deletion:", values)
else:
    print("Element not found in list.")
 
# Display final list
print("\nFinal list:", values)
```

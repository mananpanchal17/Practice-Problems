```python
#3 python to numPy array
import numpy as np
list1 = [1, 2, 3, 4, 5]
a = np.array(list1)
print("NumPy array:")
print(a)
```

```python
#2 1D array
import numpy as np
a = np.linspace(5, 50, 10)
print("10 evenly spaced values:")
print(a)
```

```python
#8 Arithmatic
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Arithmetic operations:")
print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
```

```python
#4 Reverse 1D
import numpy as np
a = np.array([1, 2, 3, 4, 5])
print("Reversed array:")
print(a[::-1])
```

```python
#10 Mean medium std derivation
import numpy as np
a = np.array([10, 20, 30, 40, 50])
print("Mean, Median and Standard Deviation:")
print("Mean:", np.mean(a))
print("Median:", np.median(a))
print("Standard deviation:", np.std(a))
```
